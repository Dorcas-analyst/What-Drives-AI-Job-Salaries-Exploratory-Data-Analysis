```python
df['posting_date'] = pd.to_datetime(df['posting_date'])
df['application_deadline'] = pd.to_datetime(df['application_deadline'])
```

```python
df.info()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['salary_usd'], kde=True, color="purple", bins = 50, edgecolor="black")
plt.title("Distribution of AI Salaries(USD)")
plt.xlabel("Salary (USD)")
plt.ylabel("Number of Jobs")
plt.savefig("Distribution of AI Salaries(USD)")
plt.show()
```

```python
currency_counts = df['salary_currency'].value_counts()
colors = plt.cm.Purples(
    0.3 + 0.7 * (currency_counts.values / currency_counts.values.max())
)
plt.bar(currency_counts.index, currency_counts.values, color=colors)
plt.xlabel('Salary Currency')
plt.ylabel('Total Observation')
plt.title('Number of Jobs by Salary Currency')
plt.savefig("Number of jobs by salary currency")
plt.show()
```

```python
sns.barplot(x="experience_level", y="salary_usd", data=df, color = "purple")
plt.title("Average Salary by Experience Level")
plt.savefig("Average Salary by Experience Level")
plt.show()
```

```python
df["skills_list"] = df["required_skills"].apply(lambda x: [s.strip() for s in x.split(",")])
```

```python
all_skills = [skill for sub in df["skills_list"] for skill in sub]
skills_counts = pd.Series(all_skills).value_counts()
```

```python
top_skills = skills_counts.head(20)
plt.figure(figsize=(10,8))
sns.barplot(x=top_skills.values, y=top_skills.index, color="purple")
plt.title("Top 20 Skills in AI Jobs")
plt.xlabel("Total Observation")
plt.ylabel("Skill")
plt.savefig("Top 20 Skills in AI Jobs")
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.violinplot(x="experience_level", y="salary_usd",  color="purple", hue="salary_currency", data=df, palette="Pastel1")
plt.title("Salary Distribution by Experience Level")
plt.savefig("Salary Distribution by Experience Level")
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(x="experience_level", y="salary_usd", color = "purple", data=df)
plt.title("Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Salary (USD)")
plt.savefig("Salary by Experience Level")
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(x="education_required", y="salary_usd", color= "purple", data=df)
plt.title("Salary by Education Level")
plt.xticks(rotation=45)
plt.savefig("Salary By Education Level")
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x="remote_ratio", y="salary_usd", color="Purple", data=df)
plt.title("Salary vs Remote Ratio")
plt.savefig("Salary vs Remote Ratio")
plt.savefig("Salary vs Remote Ratio")
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x="company_size", y="salary_usd", color="purple", data=df)
plt.title("Salary by Company Size")
plt.savefig("Salary by Company Size")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="remote_ratio", color="purple", data=df)
plt.ylabel("Total Observation")
plt.title("Remote Ratio Distribution")
plt.savefig("Remote Ratio Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="company_size", color="Purple", data = df)
plt.ylabel("Total Observation")
plt.title("Total Observation By Company Size")
plt.savefig("Total Observation By Company Size")
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.scatterplot(
    x="years_experience",
    y="salary_usd",
    hue="experience_level",
    style= "experience_level",
    palette="tab10",
    alpha=0.7,
    s=80,
    data=df
)
plt.title("Salary vs Years of Experience", fontsize=16, fontweight="bold")
plt.xlabel("Years of Experience", fontsize=12)
plt.ylabel("Salary (USD)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.3)
plt.legend(title="Experience Level", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.savefig("Salary vs Years of Experience")
plt.show()
```

```python
plt.figure(figsize=(15,6))
edu_counts = df["education_required"].value_counts()
sns.barplot(x=edu_counts.values, color="purple", y=edu_counts.index)
plt.title("Education Required for AI Jobs")
plt.show()
```

```python
numeric_cols = ["salary_usd", "remote_ratio"]
if "benefits_score" in df.columns:
    numeric_cols.append("benefits_score")
```

```python
corr = df[numeric_cols].corr()
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.savefig("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df[numeric_cols], kind="scatter", diag_kind="kde")
plt.savefig("pairplot chart")
plt.show()
```
